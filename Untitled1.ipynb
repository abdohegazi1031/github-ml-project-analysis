```python
import requests
import pandas as pd
import sqlite3
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

url = "https://api.github.com/search/repositories?q=machine+learning&sort=stars&order=desc&per_page=100"

# إضافة User-Agent لتفادي حظر GitHub API (Error 403)
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

response = requests.get(url, headers=headers)

if response.status_code == 200:
    raw_data = response.json()
    items = raw_data.get('items', [])
    print(f"Successfully retrieved {len(items)} repositories from API.")
else:
    print(f"Failed to fetch data. Status Code: {response.status_code}")
    print("Response message:", response.text)
```

```python
extracted_data = []

for repo in items:
    owner_login = repo['owner']['login'] if repo.get('owner') and 'login' in repo['owner'] else 'Unknown'
    license_name = repo['license']['name'] if repo.get('license') and 'name' in repo['license'] else 'No License'

    extracted_data.append({
        'name': repo.get('name'),
        'owner': owner_login,
        'language': repo.get('language'),
        'stargazers_count': repo.get('stargazers_count'),
        'forks_count': repo.get('forks_count'),
        'watchers_count': repo.get('watchers_count'),
        'open_issues_count': repo.get('open_issues_count'),
        'created_at': repo.get('created_at'),
        'updated_at': repo.get('updated_at'),
        'license': license_name
    })

df = pd.DataFrame(extracted_data)
print("Initial DataFrame Shape:", df.shape)
df.head(3)
```

```python

df['language'] = df['language'].fillna('Unspecified')

initial_rows = len(df)
df = df.drop_duplicates(subset=['name', 'owner']).reset_index(drop=True)
print(f"Removed {initial_rows - len(df)} duplicate records.")

print("\nRemaining Missing Values:")
print(df.isnull().sum())
```

```python

df['created_at'] = pd.to_datetime(df['created_at']).dt.date
df['updated_at'] = pd.to_datetime(df['updated_at']).dt.date

column_mapping = {
    'stargazers_count': 'stars',
    'forks_count': 'forks',
    'watchers_count': 'watchers',
    'open_issues_count': 'open_issues',
    'created_at': 'created_date',
    'updated_at': 'updated_date'
}

df = df.rename(columns=column_mapping)
print("Columns renamed successfully.")
df.head(3)
```

```python

output_filename = 'github_projects.csv'
df.to_csv(output_filename, index=False)
print(f"File successfully saved as '{output_filename}'")

df_verified = pd.read_csv(output_filename)

print("\n--- Reload Verification ---")
print(f"Verified Shape: {df_verified.shape[0]} Rows, {df_verified.shape[1]} Columns")
print("\nDataset Info:")
df_verified.info()
```

```python
conn = sqlite3.connect('github_projects.db')

query_high_stars = """
SELECT name, owner, language, stars, forks
FROM Repositories
WHERE stars > 10000
ORDER BY stars DESC;
"""

df_high_stars = pd.read_sql_query(query_high_stars, conn)

print("--- Repositories with > 10,000 Stars ---")
print(f"Total matching repositories: {len(df_high_stars)}")
display(df_high_stars.head())
```

```python
query_search_machine = """
SELECT name, owner, language, stars, forks, created_date
FROM Repositories
WHERE name LIKE '%Machine%' OR name LIKE '%machine%';
"""

df_machine_repos = pd.read_sql_query(query_search_machine, conn)

print("--- Repositories Containing 'Machine' in Name ---")
print(f"Total matching repositories: {len(df_machine_repos)}")
display(df_machine_repos)

conn.close()
```

```python
conn = sqlite3.connect('github_projects.db')

query_and_not = """
SELECT name, owner, language, stars, license
FROM Repositories
WHERE language = 'Python'
  AND stars > 20000
  AND NOT license = 'No License'
ORDER BY stars DESC;
"""

df_and_not = pd.read_sql_query(query_and_not, conn)

print("--- Query 1: Python Repositories (>20k Stars & With Valid License) ---")
print(f"Total matching records: {len(df_and_not)}")
display(df_and_not.head())
```

```python
query_complex_logic = """
SELECT name, owner, language, stars, open_issues
FROM Repositories
WHERE (language = 'Python' OR language = 'Jupyter Notebook')
  AND stars > 10000
  AND NOT open_issues > 500
ORDER BY stars DESC;
"""

df_complex_logic = pd.read_sql_query(query_complex_logic, conn)

print("--- Query 2: High Star Python/Jupyter Repositories with Managed Issues ---")
print(f"Total matching records: {len(df_complex_logic)}")
display(df_complex_logic.head())

conn.close()
```

```python
conn = sqlite3.connect('github_projects.db')

query_top10_stars = """
SELECT name, owner, language, stars, forks
FROM Repositories
ORDER BY stars DESC
LIMIT 10;
"""

df_top10 = pd.read_sql_query(query_top10_stars, conn)

print("--- Top 10 Most Popular Repositories by Stars ---")
display(df_top10)

conn.close()
```

```python
conn = sqlite3.connect('github_projects.db')

query_summary = """
SELECT
    COUNT(*) AS total_repositories,
    ROUND(AVG(stars), 2) AS average_stars
FROM Repositories;
"""

df_summary = pd.read_sql_query(query_summary, conn)

print("--- Repository Summary Statistics ---")
display(df_summary)

conn.close()
```

```python
conn = sqlite3.connect('github_projects.db')

query_language_groups = """
SELECT
    language,
    COUNT(*) AS repository_count,
    ROUND(AVG(stars), 2) AS average_stars
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 5
ORDER BY repository_count DESC;
"""

df_languages = pd.read_sql_query(query_language_groups, conn)

print("--- Languages with More Than 5 Repositories ---")
display(df_languages)

conn.close()
```

```python
conn = sqlite3.connect('github_projects.db')
query_top10 = "SELECT name, stars FROM Repositories ORDER BY stars DESC LIMIT 10;"
df_top10 = pd.read_sql_query(query_top10, conn)
conn.close()

plt.figure(figsize=(14, 7))
plt.barh(df_top10['name'][::-1], df_top10['stars'][::-1], color='skyblue')

plt.title('Top 10 Machine Learning Repositories on GitHub by Stars', fontsize=14)
plt.xlabel('Number of Stars', fontsize=12)
plt.ylabel('Repository Name', fontsize=12)
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
conn = sqlite3.connect('github_projects.db')
query_creation = "SELECT created_date FROM Repositories;"
df_creation = pd.read_sql_query(query_creation, conn)
conn.close()

df_creation['created_date'] = pd.to_datetime(df_creation['created_date'])

df_trends = df_creation.groupby(df_creation['created_date'].dt.to_period('M')).size().reset_index(name='count')

df_trends['created_date'] = df_trends['created_date'].dt.to_timestamp()

plt.figure(figsize=(10, 5))
plt.plot(df_trends['created_date'], df_trends['count'], marker='o', linestyle='-', color='teal')

plt.title('Trends in GitHub Machine Learning Repository Creation Over Time', fontsize=14)
plt.xlabel('Creation Year', fontsize=12)
plt.ylabel('Number of Repositories Created', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)

plt.gca().xaxis.set_major_locator(mdates.YearLocator())
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%Y'))

plt.tight_layout()
plt.show()
```

The final analytical synthesis evaluates repository popularity, language dominance, and historical creation trends across open-source machine learning projects to deliver actionable technical insights.

Core Popularity Patterns

Power-Law Star Distribution: Community engagement is heavily concentrated in a few foundational frameworks (tensorflow, transformers), which exceed the dataset average (~82,600 stars) by more than double.

Dual Market Demand: High star counts divide into two distinct categories: production-grade infrastructure (scikit-learn, gradio) and open-access educational roadmaps (ML-For-Beginners, awesome-machine-learning), showing equal demand for deployment tools and learning resources.

Programming Language Ecosystems

Python Monopoly: Python represents the majority of high-impact repositories, confirming its status as the default language for machine learning research and application development.

Specialized Stack Roles: Secondary languages serve specific roles—Jupyter Notebooks dominate prototyping and educational tutorials, while C++ remains essential for performance-critical deep learning backends.

Longitudinal Creation Trends

Cyclical Growth Spikes: Repository creation rates peaked sharply during major open-source breakthroughs (such as the early deep learning revolution between 2014–2015 and the transformer expansion in 2017–2018).

Maturation Phase: Creation velocity slows after initial domain spikes as the community shifts focus from creating new projects to maintaining established, enterprise-standard frameworks.

Strategic Business Takeaways

Licensing Accelerates Adoption: The vast majority of top-tier projects use permissive open-source licenses (MIT and Apache 2.0), which minimize legal risk and encourage rapid commercial adoption.

Project Health Indicator: Top repositories sustain high fork counts alongside controlled open issue ratios, demonstrating that active community contribution is vital for long-term project viability.
